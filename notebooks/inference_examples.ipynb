## 1. Import Dependencies

```python
import os

import torch
from dotenv import load_dotenv
from transformers import (
    AutoModelForSeq2SeqLM,
    AutoTokenizer,
)
from transformers.models.mbart50.tokenization_mbart50 import FAIRSEQ_LANGUAGE_CODES
from transformers import GenerationConfig
```

## 2. Make inference to mbart

```python
MODEL_ID = "Iscte-Sintra/mbart-51-stable-v1.0"
SRC_LANG = "kea_st_XX"
TGT_LANG = "pt_XX"

# Example Kriolu sentence
TEXT = "N sta dretu."

load_dotenv()
HF_TOKEN = os.getenv("HF_TOKEN")
if not HF_TOKEN:
    raise EnvironmentError(
        "HF_TOKEN was not found. "
        "Make sure you have HF_TOKEN=... in your .env file."
    )

tok = AutoTokenizer.from_pretrained(MODEL_ID)
model = AutoModelForSeq2SeqLM.from_pretrained(MODEL_ID)

GEN = GenerationConfig(num_beams=3, max_new_tokens=64)
tok.src_lang = SRC_LANG
tok.tgt_lang = TGT_LANG
enc = tok(TEXT, return_tensors="pt")
out = model.generate(**enc, generation_config=GEN,
                         forced_bos_token_id=tok.lang_code_to_id[TGT_LANG])  # REQUIRED on mBART-50
print(tok.batch_decode(out, skip_special_tokens=True)[0])
```
